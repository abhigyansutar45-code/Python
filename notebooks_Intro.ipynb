```python
# Python Calculator

print("===== CALCULATOR =====")

a = float(input("Enter your first number: "))
operator = input("Enter operator (+, -, *, /): ")
b = float(input("Enter your second number: "))

if operator == "+":
    print("Result =", a + b)

elif operator == "-":
    print("Result =", a - b)

elif operator == "*":
    print("Result =", a * b)

elif operator == "/":
    if b != 0:
        print("Result =", a / b)
    else:
        print("Error: Cannot divide by zero!")

else:
    print("Invalid operator!")
```
